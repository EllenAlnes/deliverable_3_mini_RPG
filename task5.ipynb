## Battle with weapons

```python
sword = Weapon("sword", Dice(7))

hero = Warrior("Per", weapon=sword)
enemy = Mage("Hans")

battle = Battle(hero, enemy)
hero_life, enemy_life = battle.simulate_battle()
```
